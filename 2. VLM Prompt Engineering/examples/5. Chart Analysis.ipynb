# 5. 차트 및 그래프 분석 — 막대그래프 / 선그래프 / 대시보드

VLM은 차트의 **정확한 픽셀 단위 수치보다는 상대적 비교(순위, 추세)**에
강합니다. 어떤 질문이 잘 통하는지 직접 확인해봅니다.

```python
from vlm_utils import ask, show

BAR_CHART = "data/bar_chart.png"
LINE_CHART = "data/line_chart.png"
DASHBOARD = "data/dashboard.png"
```

## 막대그래프(Bar Chart) 분석
정답: 최댓값 5월(820), 최솟값 2월(310)

```python
show(BAR_CHART)
bar_qa = ask(
    BAR_CHART,
    "이 막대그래프에서 값이 가장 큰 항목과 가장 작은 항목을 알려주고, "
    "두 값의 차이를 추정해줘. 축 단위도 함께 확인해서 답변에 포함해.",
)
print(bar_qa)
```

## 선그래프(Line Chart) 분석
정답: 7일차 이후 급격히 하락(8일차 부근에서 저점)

```python
show(LINE_CHART)
line_qa = ask(
    LINE_CHART,
    "이 선그래프의 전체적인 추세를 설명하고, 값이 급격히 변한 구간이 "
    "있다면 x축 기준으로 어디인지 짚어줘.",
)
print(line_qa)
```

## 대시보드 분석
여러 차트가 한 화면에 있을 때는 **한 번에 다 묻기보다 차트별로 나눠 질의**하면
정확도가 올라갑니다. 먼저 전체 인상을 물어본 뒤, 세부 차트를 짚어 재질의합니다.

```python
show(DASHBOARD, width=560)
overview = ask(
    DASHBOARD,
    "이 대시보드에서 지난 달 대비 가장 크게 변한 지표를 찾아, "
    "어느 차트에서 확인했는지와 함께 설명해줘.",
)
print(overview)
```

```python
pie_only = ask(
    DASHBOARD,
    "대시보드 왼쪽 아래에 있는 원형 차트(고객 구성 비율)만 보고, "
    "가장 비중이 큰 항목과 그 비율을 알려줘.",
)
print(pie_only)
```
