# 차트 및 그래프 분석 — 막대그래프 / 선그래프 / 대시보드

VLM은 차트의 **정확한 픽셀 단위 수치보다는 상대적 비교(순위, 추세)**에
강합니다. 어떤 질문이 잘 통하는지 직접 확인해봅니다.

```python
from vlm_utils import ask, show

BAR_CHART = "data/bar_chart.png"
LINE_CHART = "data/line_chart.png"
DASHBOARD = "data/dashboard.png"
```

## 막대그래프(Bar Chart) 분석
정답: 최댓값 5월(820), 최솟값 2월(310)

```python
show(BAR_CHART)
```

```python
bar_qa = ask(
    BAR_CHART,
    "이 막대그래프에서 값이 가장 큰 항목과 가장 작은 항목을 알려주고, "
    "두 값의 차이를 추정해줘. 축 단위도 함께 확인해서 답변에 포함해.",
)
print(bar_qa)
```

```python
BAR_CHART01 = "./data/bar_chart01.png"
show(BAR_CHART01)
```

```python
bar_qa = ask(
    BAR_CHART01,
    "2026년에서 평균적으로 가장 인기가 많은 음악과 적은 음악을 찾아줘."
    "추가로 가장 음악을 많이 듣는 월과 가장 음악을 적게 듣는 월도 찾아줘.",
)
print(bar_qa)
```

## 선그래프(Line Chart) 분석

```python
line_qa = ask(
    LINE_CHART,
    "이 선그래프의 전체적인 추세를 설명하고, 값이 급격히 변한 구간이 있다면, "
    "x축 기준으로 어디인지 짚어줘.",
)
print(line_qa)
```

```python
LINE_CHART01 = "./data/line_chart01.png"
show(LINE_CHART01)
```

```python
line_qa = ask(
    LINE_CHART01,
    "이 선그래프의 전체적인 추세를 설명하고, "
    "어느 년도에 중국이 미국을 추월하게 되었는지 알려줘.",
)
print(line_qa)
```

## 대시보드 분석
여러 차트가 한 화면에 있을 때는 **한 번에 다 묻기보다 차트별로 나눠 질의**하면 정확도가 올라갑니다. 

```python
show(DASHBOARD, width=560)
```

```python
pie_only = ask(
    DASHBOARD,
    "대시보드 왼쪽 아래에 있는 원형 차트(고객 구성 비율)만 보고, "
    "가장 비중이 큰 항목과 그 비율을 알려줘.",
)
print(pie_only)
```

```python
DASHBOARD01 = "./data/dashboard01.png"
show(DASHBOARD01, width=560)
```

```python
left_only = ask(
    DASHBOARD01,
    "대시보드 왼쪽 유업종 비교 분석과 재무제표를 분석한 후, "
    "어떤 업종에 투자하는게 좋은지 알려줘.",
)
print(left_only)
```
