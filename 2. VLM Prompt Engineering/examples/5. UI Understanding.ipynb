# 6. UI 이해 — 웹페이지 분석 / Figma 화면 이해 / UX 분석

화면 스크린샷·디자인 목업을 VLM으로 분석해 **레이아웃 구조 파악, 디자인
스펙 문서화, 사용성 피드백**까지 이어지는 흐름을 실습합니다.

```python
from vlm_utils import ask, show

WEBPAGE = "data/webpage.png"
FIGMA_MOCK = "data/figma_mock.png"
```

## 웹페이지 분석 (스크린샷 이해)

```python
show(WEBPAGE, width=560)
nav_qa = ask(
    WEBPAGE,
    "이 웹페이지 스크린샷에서 주요 내비게이션 메뉴 항목을 나열하고, "
    "가장 눈에 띄는 CTA(Call To Action) 버튼이 어디에 있고 어떤 문구인지 설명해줘.",
)
print(nav_qa)
```

## Figma 화면 이해
디자인 시안(목업)에서 컴포넌트 구성을 뽑아내면 디자인 스펙 문서화를
자동화하는 데 참고할 수 있습니다. (정확한 색상 코드·픽셀 값은 참고용으로만 활용)

```python
show(FIGMA_MOCK, width=320)
components_qa = ask(
    FIGMA_MOCK,
    "이 화면 캡처에서 화면을 구성하는 주요 컴포넌트(상단 타이틀, 프로필, "
    "리스트 항목, 하단 내비게이션 등)를 나열하고 전체적인 레이아웃 구조를 설명해줘.",
)
print(components_qa)
```

## UX 분석
화면 구조를 바탕으로 **사용성 관점의 피드백**을 요청하는 상위 태스크입니다.
"어떤 사용자 기준으로 평가할지"를 명시하면 답변이 훨씬 구체화됩니다.

```python
ux_qa = ask(
    FIGMA_MOCK,
    "이 화면을 처음 사용해보는 사용자 입장에서, 원하는 메뉴를 직관적으로 "
    "찾을 수 있는지 평가해줘. 개선이 필요하다면 구체적인 이유와 함께 제안해줘.",
)
print(ux_qa)
```
